```python
import datetime

class Option:
    def __init__(self, symbol, option_type, strike_price, expiration_date):
        self.symbol = symbol
        self.option_type = option_type
        self.strike_price = strike_price
        self.expiration_date = expiration_date

    def __str__(self):
        return f"{self.symbol} {self.option_type} {self.strike_price} {self.expiration_date}"

def generate_option_chains(symbol, expiration_dates, strike_prices):
    options = []

    # Generate call options
    for expiration_date in expiration_dates:
        for strike_price in strike_prices:
            call_option = Option(symbol, "Call", strike_price, expiration_date)
            options.append(call_option)

    # Generate put options
    for expiration_date in expiration_dates:
        for strike_price in strike_prices:
            put_option = Option(symbol, "Put", strike_price, expiration_date)
            options.append(put_option)

    return options

# Define the symbol, expiration dates, and strike prices
symbol = "AAPL"
expiration_dates = [
    datetime.date(2023, 5, 19),
    datetime.date(2023, 6, 16),
    datetime.date(2023, 7, 21)
]
strike_prices = [120, 130, 140]

# Generate the option chains
option_chains = generate_option_chains(symbol, expiration_dates, strike_prices)

# Print the option chains
for option in option_chains:
    print(option)
```
