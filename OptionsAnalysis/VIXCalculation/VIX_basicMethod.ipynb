```python
import math

# Define the inputs for the calculation
sigma = 0.2  # Implied volatility of the stock's at-the-money options
t = 30  # Time to expiration of the options in days (assuming 30 days)

# Perform the calculation
vix_ind = 100 * sigma * math.sqrt(t/365)

# Print the result
print(f"VIX-style index for individual stock: {vix_ind:.2f}")
```
