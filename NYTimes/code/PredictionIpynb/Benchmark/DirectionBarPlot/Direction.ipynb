```python
# multiplicate element-wise, and count the positivie elements
def count_pos(series):
    count = 0
    for x in series:
        if x>0:
            count += 1
    return count
```

```python
plt.rcParams['font.size'] = 13

percentage = []
for i in range(len(direction_counts)):
    pct = (direction_counts[i] / total*100)
    percentage.append(round(pct,2))
data=pd.DataFrame(data=percentage).rename(columns={0:"Percentage"})

# depict illustration
# colors_list = ['Red','Orange', 'Blue', 'Purple','Orange', 'Blue','Yellow']
fig, ax = plt.figure(figsize=(20,7)), plt.axes()

graph = ax.bar(Names,data.Percentage.to_list(), color=(0.2, 0.4, 0.6, 0.6), edgecolor='blue')
plt.title('Percentage of days in which the model predicted the right direction (up or down)')

i = 0
for p in graph:
    width = p.get_width()
    height = p.get_height()
    x, y = p.get_xy()
    plt.text(x+width/2,
             y+height*1.01,
             str(data.Percentage[i])+'%',
             ha='center',
             weight='bold')
    i+=1

plt.tight_layout()
plt.show()
```
