```python
import pandas as pd
import numpy as np
import seaborn as sns
import statistics
import matplotlib.pyplot as plt
sns.set(font_scale=1.5)

Type="Quantile"
df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/NYTimes/data/pkl/df_dates.pkl")
df = df[["VIX","NewsVolume","VIX return", "NewsVolume return"]]
```

```python
df["VIX return"].mean()
```

```python
df["VIX NR"] = df["VIX return"].mean()
df["NewsVolume NR"] = df["NewsVolume return"].mean()

df["VIX AR"] = df["VIX return"] - df["VIX NR"]
df["NewsVolume AR"] = df["NewsVolume return"] - df["NewsVolume NR"]
```

```python
df["VIX AR"]
```

# 1) STDEV

```python
vix_stdev = 2*statistics.stdev(df["VIX AR"].dropna())
vix_mean = df["VIX AR"].mean()
vix = df.loc[(df["VIX AR"] >= vix_mean+vix_stdev) | (df["VIX AR"] <= vix_mean-vix_stdev)]
x0 = vix_mean-vix_stdev
x1 =vix_mean+vix_stdev
plt.title("The New York Times\nTwo Standard diviations\nabove or below the Carhart AR")
# PLOT:
len(vix)

ax = sns.distplot(df['VIX AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()

#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')

ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import statistics

# Define new color values
all_news_color = "#4CAF50"  # Green
filtered_news_color = "#F44336"  # Red
highlighted_area_color = "#FFCDD2"  # Light red
vertical_lines_color = "#E57373"  # Lighter red

# Calculate volume mean and standard deviation
volume_stdev = 2 * statistics.stdev(df["NewsVolume AR"].dropna())
volume_mean = df["NewsVolume AR"].mean()

# Filter the data for volumes outside of two standard deviations
volume = df.loc[(df["NewsVolume AR"] >= volume_mean + volume_stdev) | 
                     (df["NewsVolume AR"] <= volume_mean - volume_stdev)]

# Set up variables for plotting
x0 = volume_mean - volume_stdev
x1 = volume_mean + volume_stdev
sns.set(font_scale=1.5)

# Create the plot
fig, ax = plt.subplots(figsize=(10, 6))
sns.distplot(df["NewsVolume AR"], hist=True, color=all_news_color, ax=ax, label="All News Volume")
sns.distplot(volume["NewsVolume AR"], hist=True, color=filtered_news_color, ax=ax, label="Filtered News Volume")

# Highlight the filtered area
kde_x, kde_y = ax.lines[-1].get_data()
ax.fill_between(kde_x, kde_y, where=(kde_x < x0) | (kde_x > x1), interpolate=True, color=highlighted_area_color)

# Add vertical lines and legend
ax.axvline(x0, color=vertical_lines_color, label="2 std below mean")
ax.axvline(x1, color=vertical_lines_color, label="2 std above mean")
ax.legend()
ax.set_xlabel("News Volume")
ax.set_ylabel("Density")
ax.set_title("The New York Times\nNews Volume Two Standard Deviations Above or Below the Mean AR")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import statistics

sns.set_style('white')
sns.set_palette('dark')

volume_stdev = 2 * statistics.stdev(df["NewsVolume AR"].dropna())
volume_mean = df["NewsVolume AR"].mean()
volume = df.loc[(df["NewsVolume AR"] >= volume_mean + volume_stdev) | (df["NewsVolume AR"] <= volume_mean - volume_stdev)]
x0 = volume_mean - volume_stdev
x1 = volume_mean + volume_stdev
volume_color = "#C44E52"

plt.figure(figsize=(8, 6))
plt.title("The New York Times\nDistribution of News Volume Abnormal Returns\n(Two Standard Deviations Above or Below the Mean)", fontsize=16)
plt.xlabel("News Volume Abnormal Returns", fontsize=14)
plt.ylabel("Density", fontsize=14)

ax = sns.distplot(df['NewsVolume AR'], hist=True, kde=True, color='#388E3C', kde_kws={'linewidth': 2})
kde_x, kde_y = ax.lines[0].get_data()

p1 = plt.axvline(x=x0, color=volume_color, label="Mean $\pm$ 2 SD", linewidth=2)
p2 = plt.axvline(x=x1, color=volume_color, linewidth=2)
ax.fill_between(kde_x, kde_y, where=(kde_x < x0) | (kde_x > x1), interpolate=True, color=volume_color, alpha=0.5)
plt.legend(fontsize=12)

sns.despine()

plt.show()
```
