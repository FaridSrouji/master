```python
# Columns to plot 
fig, axes = plt.subplots(4,2,figsize=(30,45))
pr_count = df["PressReleaseCount"].loc[(df["Hour"]>=0) & (df["Hour"]<4)].count()
pr_num = df["PressReleaseCount"].loc[(df["Hour"]>=0) & (df["Hour"]<4)].sum()

fig.suptitle("Positive Sentiment\nHours 0 to 3\nTotal hours with Press Releases"+str(int(pr_count)) +"Total Press Releases: "+str(int(pr_num)) , fontsize=30)
fig.subplots_adjust(top=1.88)


# fig = plt.gcf()
fig.tight_layout(pad=5.8)
sns.set(font_scale = 1.5)

for i in range(4):
    print("i="+str(i))
    vix_return = df["vix_ln_return"].loc[df["Hour"]==i]
    pos_return = df["pos_ln_return"].loc[df["Hour"]==i]
    countObsVix = len(vix_return)
    countObspos = len(pos_return)
    print("pos_return="+str(countObsVix)+" vix_ln_return="+str(countObspos))
    sns.regplot(pos_return,vix_return, data = df, ax=axes[i,0]).set_title("Hour:"+str(i)+"\nTotal Observations:"+str(countObspos), fontsize=24) # regression plot
    sns.kdeplot(pos_return,vix_return,data = df,shade=True, ax=axes[i,1]).set_title("Hour:"+str(i)+"\nTotal Observations:"+str(countObspos), fontsize=24) # kernel density estimate plot

plt.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images_press_release/KdeAndRegPlot/KdeAndRegPlotHourly_pos_0-3.png", dpi=216)
plt.show()
```
