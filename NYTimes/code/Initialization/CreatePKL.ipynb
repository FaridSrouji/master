```python
import pandas as pd
df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/NYTimes/data/pkl/TimeSentiment.pkl")
```

```python
df.sort_values(by="Date")
```

```python
dates = df.Date.unique()
dates.sort()
# del df
df_dates = pd.DataFrame(index =dates, columns=["Date", "headline", "lead_paragraph", "NewsCount"])

df_dates.Date = dates
df_dates.headline = ""
df_dates.lead_paragraph = ""
df_dates.NewsCount = 0

print(df_dates)
```

```python
total=0

for i, row in df.T.iteritems(): 
    # print(hour)
    date = row["Date"]  
    #print(date)
    articles = df_dates.at[date, "headline"]
    titles = df_dates.at[date, "lead_paragraph"]
    articles

    df_dates.at[date, "headline"] = articles + ". " + row["headline"]
    df_dates.at[date, "lead_paragraph"] = titles + ". " + row["lead_paragraph"]
    df_dates.at[date, "NewsCount"] +=1
    total +=1

print(total)
```

```python
df_dates["WordCount"] = df_dates["lead_paragraph"].apply(lambda x:len(str(x).split()))
```
