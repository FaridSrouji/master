```python
import pandas as pd 
import numpy as np
df_time = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/NYTimes/data/pkl/TimeSentiment.pkl")
```

```python
df_time['Hour'] = df_time['Hour'].astype(int)
df_time.sort_values(by="Date",inplace = True)
```

```python
# df_time.loc[df_time["Date"] >"2011-01-02"]
```

```python
df_time.sort_values(by="Time")
```

```python
dates=df_time.Date.unique()

df = pd.DataFrame(columns=["DateHour", "Hour", "lead_paragraph", "headline", "NewsCount"])
df_hour = pd.DataFrame(index =dates, columns=["DateHour", "Hour", "lead_paragraph", "headline", "NewsCount"])
df_hour.index = dates

df.DateHour = ""
# df.Hour = ""
df.Articles = ""
df.Title = ""
df.NewsCount = 0

df_hour.DateHour = ""
# df_hour.Hour = ""
df_hour.headline = ""
df_hour.lead_paragraph = ""
df_hour.NewsCount = 0

for h in range(24):
    for i, row in df_time.loc[df_time["Hour"]==h].T.iteritems():
        hour = row["Hour"]
        date = row["Date"]  
        df_hour.at[date, "Hour"] = h
        df_hour.at[date, "DateHour"] = date + "_" + str(h)
        # if (hour==h):
        print("Hour="+str(h))
        df_hour.at[date, "lead_paragraph"] = df_hour.at[date, "lead_paragraph"] + ". " + row["lead_paragraph"]
        df_hour.at[date, "headline"] = df_hour.at[date, "headline"] + ". " + row["headline"]
        df_hour.at[date, "NewsCount"] +=1

    frames = [df, df_hour]
    df = pd.concat(frames)
    del df_hour
    df_hour = pd.DataFrame(index =dates, columns=["DateHour", "Hour", "lead_paragraph", "headline", "NewsCount"])
    df_hour.DateHour = ""
    df_hour.Hour = ""
    df_hour.lead_paragraph = ""
    df_hour.headline = ""
    df_hour.NewsCount = 0
```

```python
df.dropna(inplace=True)
```

```python
df["WordCount"] = df["lead_paragraph"].apply(lambda x:len(str(x).split()))
```
