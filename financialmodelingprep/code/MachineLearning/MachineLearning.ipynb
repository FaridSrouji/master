```python
conditions = [
    (df_sentiment['pos'] < df_sentiment['neg']),
    (df_sentiment['pos'] > df_sentiment['neg_ln_return']),
    ((df_sentiment['pos'] == 0) & (df_sentiment['neg_ln_return'] == 0)),
    ((df_sentiment['pos'] == df_sentiment['neg_ln_return']) & ((df_sentiment['pos_ln_return'] != 0)))
    ]

# create a list of the values we want to assign for each condition
values = ['Negative', 'Positive', 'Neutral', 'Equal']

# create a new column and use np.select to assign values to it using our lists as arguments
df_sentiment['SentimentValue'] = np.select(conditions, values)
```
