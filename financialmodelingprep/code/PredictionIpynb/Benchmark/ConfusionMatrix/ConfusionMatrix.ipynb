```python
# direction arrays
directions_VIX = target_VIX - target_VIX.shift()
directions_random = predictions_random - predictions_random.shift()
directions_lr = predictions_lr - predictions_lr.shift()
directions_lor = predictions_lor - predictions_lor.shift()
directions_rf = predictions_rf - predictions_rf.shift()
directions_ml = predictions_ml -  predictions_ml.shift()
directions_hv = predictions_hv - predictions_hv.shift()
directions_nn = predictions_nn - predictions_nn.shift()
directions_lstm = predictions_lstm - predictions_lstm.shift()
directions_gru = predictions_gru - predictions_gru.shift()
```

```python
# multiplicate element-wise, and count the positivie elements
def count_pos(series):
    count = 0
    for x in series:
        if x>0:
            count += 1
    return count
```
