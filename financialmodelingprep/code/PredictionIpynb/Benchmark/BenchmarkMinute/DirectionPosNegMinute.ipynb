```python
def count_pos_neg(model, VIX, totPositive=0,totNegative=0, totPosNeg=0, totNegPos=0,totZero=0):
    for (val1, val2) in zip(model, VIX):
        if (val1>0 and val2>0):
            totPositive = totPositive+1
        elif(val1<0 and val2<0):
            totNegative = totNegative+1
        elif (val1==0 and val2==0):
            totZero = totZero+1
        elif(val1>0 and val2<0):
            totPosNeg = totPosNeg+1
        elif(val1<0 and val2>0):
            totNegPos = totNegPos+1
            print("ERROR")
    return totPositive, totNegative, totZero
```
