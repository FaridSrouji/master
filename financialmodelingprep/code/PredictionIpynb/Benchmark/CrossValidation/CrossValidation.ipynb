```python
import pandas as pd
from sklearn.model_selection import TimeSeriesSplit
from sklearn.linear_model import LinearRegression

# Load time-series data
# df = pd.read_csv('time_series_data.csv')

# Define number of splits for time-series cross-validation
n_splits = 5

# Define features and target variable
X = df[['Sentiment return', 'Negative return', 'Positive return']].dropna()
y = df['VIX return'].dropna()

# Initialize the time-series cross-validation splitter
tscv = TimeSeriesSplit(n_splits=n_splits)

# Initialize the model
model = LinearRegression()

# Loop through each fold in the time-series cross-validation
for train_index, test_index in tscv.split(X):
    # Split the data into training and validation sets for this fold
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    # Fit the model to the training data for this fold
    model.fit(X_train, y_train)

    # Evaluate the model on the validation data for this fold
    score = model.score(X_test, y_test)

    # Print the score for this fold
    print(f"Score for fold {i}: {score}")
```
