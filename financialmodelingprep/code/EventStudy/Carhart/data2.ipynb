```python
import yfinance as yf
import numpy as np
spx_data = yf.download('^GSPC', start='2010-01-01', end='2023-03-05')
vix_data = yf.download('^VIX', start='2010-01-01', end='2023-03-05')
spx_returns = np.log(spx_data['Adj Close'] / spx_data['Adj Close'].shift(1))
vix_returns = np.log(vix_data['Adj Close'] / vix_data['Adj Close'].shift(1))
VIX_skew = 100 * (vix_data['Close']**2 - (spx_data['Close'] * np.exp(vix_returns * np.sqrt(252)))**2) / (spx_data['Close'] * np.exp(vix_returns * np.sqrt(252)))**2
VIX_skew.plot()
```
