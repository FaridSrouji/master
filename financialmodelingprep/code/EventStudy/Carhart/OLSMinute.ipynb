```python
X = df["Sentiment return"].values

y = df["VIX return"].values

import statsmodels.api as sm
import numpy as np
# Fit the Carhart model
model = sm.OLS(y, sm.add_constant(X)).fit()

# Compute the predicted VIX returns
y_pred = model.predict(sm.add_constant(X))

# Compute the residuals
residuals = y - y_pred

# Compute the normal returns
mu = np.mean(residuals)
sigma = np.std(residuals)
normal_returns = (residuals - mu) / sigma
```

```python
# Define event windows
event_window_size = 5 # days
event_window_starts = df2[df2['Sentiment return'] > 0.5].index
event_windows = [df2.loc[start:start + pd.Timedelta(days=event_window_size - 1)].index 
                 for start in event_window_starts]
event_windows = [window for window in event_windows if len(window) == event_window_size]

# Calculate returns
returns = []
for window in event_windows:
    print(window)
    window_vix_data = df.loc[window]
    window_returns = window_vix_data.iloc[-1]["VIX returns"] / window_vix_data.iloc[0]["VIX returns"] - 1
    returns.append(window_returns)

# Calculate abnormal returns
x = df2['Sentiment return'].values.reshape(-1, 1)
y = df2["VIX return"].values.reshape(-1, 1)
model = sm.OLS(y, sm.add_constant(x)).fit()
abnormal_returns = []
for window_return, window in zip(returns, event_windows):
    expected_return = model.predict([[1, df2.loc[window[0], 'Sentiment return']]])[0]
    abnormal_return = window_return - expected_return
    abnormal_returns.append(abnormal_return)

# Aggregate abnormal returns
aggregate_abnormal_return = sum(abnormal_returns)
```
