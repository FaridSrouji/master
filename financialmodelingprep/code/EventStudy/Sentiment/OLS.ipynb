```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Load the data
df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/df.pkl")
df_dates = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/code/EventStudy/PKL/Sentiment2stdev.pkl")
# df_selected = df[df['Date'].isin(df_dates['Date'])]
df_selected = df[~df['Date'].isin(df_dates['Date'])]

# print the selected dataframe
print(df_selected)
```

```python
df_selected = df_selected[1:]
```

```python
# Define the dependent and independent variables
y = df_selected[['VIX return']]
x = df_selected[['NewsVolume return']]

# Add a constant term to the independent variables
# x = sm.add_constant(x)

# Fit the OLS regression model
model1 = sm.OLS(y, x).fit()

# Print the model summary
print(model1.summary())
```

```python
df_selected = df[df['Date'].isin(df_dates['Date'])]

# print the selected dataframe
print(df_selected)
```

```python
# Define the dependent and independent variables
y = df_selected[['VIX return']]
x = df_selected[['NewsVolume return']]

# Add a constant term to the independent variables
# x = sm.add_constant(x)

# Fit the OLS regression model
model2 = sm.OLS(y, x).fit()

# Print the model summary
print(model2.summary())
```

```python
model1.params
```

```python
# Combine summary tables for both models into a single table
summary_table = pd.concat([pd.DataFrame({'Model': 'Events Window', 
                                          'Coefficient': model1.params.apply(lambda x: f'{x:.3f}'), 
                                          'Standard Error': model1.bse.apply(lambda x: f'{x:.3f}'), 
                                          't-value': model1.tvalues.apply(lambda x: f'{x:.3f}'), 
                                          'p-value': model1.pvalues.apply(lambda x: f'{x:.3f}')}), 
                           pd.DataFrame({'Model': 'Estimation Window', 
                                          'Coefficient': model2.params.apply(lambda x: f'{x:.3f}'), 
                                          'Standard Error': model2.bse.apply(lambda x: f'{x:.3f}'), 
                                          't-value': model2.tvalues.apply(lambda x: f'{x:.3f}'), 
                                          'p-value': model2.pvalues.apply(lambda x: f'{x:.3f}')})],
                          ignore_index=True)

# Set the order of the columns in the summary table
summary_table = summary_table[['Model', 'Coefficient', 'Standard Error', 't-value', 'p-value']]
```

```python
summary_table
```

```python

import pandas as pd
import docx
# Create a new Word document
doc = docx.Document()

# Add a heading
doc.add_heading('Summary Statistics', level=1)

# Add a table with the summary statistics
table = doc.add_table(rows=len(summary_table)+1, cols=len(summary_table.columns))
# Add the column headings
for i, column_name in enumerate(summary_table.columns):
    table.cell(0, i).text = column_name
# Add the data rows
for i, index_label in enumerate(summary_table.index):
    for j, column_name in enumerate(summary_table.columns):
        table.cell(i+1, j).text = str(summary_table.loc[index_label, column_name])

# Save the document to a file
doc.save('summary_stats.docx')
```
