```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Load the data
df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/df.pkl")
df_dates = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/code/EventStudy/PKL/volume2stdev.pkl")
# df_selected = df[df['Date'].isin(df_dates['Date'])]
df_selected = df[~df['Date'].isin(df_dates['Date'])]

# print the selected dataframe
print(df_selected)
```

```python
df_selected = df_selected[1:]
```

```python
# Define the dependent and independent variables
y = df_selected[['VIX return']]
x = df_selected[['NewsVolume return']]

# Add a constant term to the independent variables
# x = sm.add_constant(x)

# Fit the OLS regression model
model = sm.OLS(y, x).fit()

# Print the model summary
print(model.summary())
```
