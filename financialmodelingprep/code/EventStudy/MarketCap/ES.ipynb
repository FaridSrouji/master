```python
import scipy.signal.signaltools

def _centered(arr, newsize):
    # Return the center newsize portion of the array.
    newsize = np.asarray(newsize)
    currsize = np.array(arr.shape)
    startind = (currsize - newsize) // 2
    endind = startind + newsize
    myslice = [slice(startind[k], endind[k]) for k in range(len(endind))]
    return arr[tuple(myslice)]

scipy.signal.signaltools._centered = _centered

import eventstudy as es
import numpy as np
import matplotlib.pyplot as plt

import pandas as pd
import pandas_datareader as web
```

```python
import pandas as pd
import docx
from eventstudy import EventStudy

# Load example data
df = pd.read_csv('https://raw.githubusercontent.com/lemairejean-baptiste/eventstudy/main/eventstudy/example_data.csv')

# Define event dates and window
event_dates = ['2022-01-01', '2022-02-01', '2022-03-01']
event_window = [-10, 10]

# Instantiate EventStudy object
event_study = EventStudy(df, event_dates, event_window)

# Run event study
event_study.run()

# Get ARs, CARs, standard errors, t-statistics, p-values, and confidence intervals
ars = event_study.abnormal_returns
cars = event_study.cumulative_abnormal_returns
se = event_study.standard_errors
t_stats = event_study.t_statistics
p_vals = event_study.p_values
conf_ints = event_study.confidence_intervals

# Combine results into single data frame
results = pd.concat([ars, cars, se, t_stats, p_vals, conf_ints], axis=1)
results.columns = ['AR', 'CAR', 'SE', 't-stat', 'p-value', 'CI']

# Create Word document
doc = docx.Document()

# Add table to Word document
table = doc.add_table(rows=1, cols=len(results.columns))
hdr_cells = table.rows[0].cells
for i, col in enumerate(results.columns):
    hdr_cells[i].text = col
for index, row in results.iterrows():
    row_cells = table.add_row().cells
    for i, val in enumerate(row):
        row_cells[i].text = '{:.3f}'.format(val)

# Save Word document
doc.save('event_study_results.docx')
```
