```python
import pandas as pd
import numpy as np
import seaborn as sns
import statistics
import matplotlib.pyplot as plt
sns.set(font_scale=1.5)

Type="Quantile"
df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/df.pkl")
df = df[["VIX","NewsVolume","VIX return", "NewsVolume return"]]
```

```python
df["VIX return"].mean()
```

```python
df["VIX NR"] = df["VIX return"].mean()
df["NewsVolume NR"] = df["NewsVolume return"].mean()

df["VIX AR"] = df["VIX return"] - df["VIX NR"]
df["NewsVolume AR"] = df["NewsVolume return"] - df["NewsVolume NR"]
```

```python
df["VIX AR"] = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/code/EventStudy/PKL/AR_Carhart.pkl")
df["VIX NR"] = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/code/EventStudy/PKL/NR_Carhart.pkl")
```

# 1) STDEV

```python
vix_stdev = 1.5*statistics.stdev(df["VIX AR"].dropna())
vix_mean = df["VIX AR"].mean()
vix = df.loc[(df["VIX AR"] >= vix_mean+vix_stdev)]# | (df["VIX AR"] <= vix_mean-vix_stdev)]
x0 = vix_mean-vix_stdev
x1 =vix_mean+vix_stdev
plt.title("Press Release\nTwo Standard diviations\nabove or below the Carhart AR")
# PLOT:
len(vix)

ax = sns.distplot(df['VIX AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()

#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')

ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import statistics

sns.set_style('white')
sns.set_palette('dark')

volume_stdev = 2 * statistics.stdev(df["NewsVolume AR"].dropna())
volume_mean = df["NewsVolume AR"].mean()
volume = df.loc[(df["NewsVolume AR"] >= volume_mean + volume_stdev) | (df["NewsVolume AR"] <= volume_mean - volume_stdev)]
x0 = volume_mean - volume_stdev
x1 = volume_mean + volume_stdev
volume_color = "#C44E52"

plt.figure(figsize=(8, 6))
plt.title("Press-Release\nDistribution of News Volume Abnormal Returns\n(Two Standard Deviations Above or Below the Mean)", fontsize=16)
plt.xlabel("News Volume Abnormal Returns", fontsize=14)
plt.ylabel("Density", fontsize=14)

ax = sns.distplot(df['NewsVolume AR'], hist=True, kde=True, color='#388E3C', kde_kws={'linewidth': 2})
kde_x, kde_y = ax.lines[0].get_data()

p1 = plt.axvline(x=x0, color=volume_color, label="Mean $\pm$ 2 SD", linewidth=2)
p2 = plt.axvline(x=x1, color=volume_color, linewidth=2)
ax.fill_between(kde_x, kde_y, where=(kde_x < x0) | (kde_x > x1), interpolate=True, color=volume_color, alpha=0.5)
plt.legend(fontsize=12)

sns.despine()

plt.show()
```
