```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set the decay factor
decay_factor = 0.9  # Adjust the decay factor as needed

# Compute the weights
weights = np.power(decay_factor, np.arange(window_size, 0, -1))

# Compute the moving average with decaying weights
moving_average_decay = np.convolve(df["PressReleaseVolume AR"], weights, mode='valid')

# Create an empty DataFrame for the event study results
event_study = pd.DataFrame(data = df_dates,columns=["Event Date", "Abnormal Return"])

# Iterate over the event dates
for event_date in df_dates.values:
    # Compute the start and end indices for the event window
    print(event_date)
    start_index = df.index.get_loc(event_date) - window_size + 1
    end_index = df.index.get_loc(event_date) + 1
    
    # Slice the data for the event window
    event_data = df.iloc[start_index:end_index]
    
    # Compute the average of the event window
    event_average = event_data["PressReleaseVolume AR"].mean()
    
    # Compute the abnormal return
    abnormal_return = event_average - np.mean(moving_average_decay)
    
    # Append the event date and abnormal return to the event study DataFrame
    event_study = event_study.append({"Event Date": event_date, "Abnormal Return": abnormal_return}, ignore_index=True)

# Plot the event study
plt.figure(figsize=(8, 6))
plt.plot(event_study["Event Date"], event_study["Abnormal Return"], marker="o")
plt.axhline(y=0, color="black", linestyle="--")
plt.title("Event Study: Abnormal Returns based on Moving Average with Decaying Weights")
plt.xlabel("Event Date")
plt.ylabel("Abnormal Return")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import statistics

sns.set_style('white')
sns.set_palette('dark')

# Compute moving average
window_size = 5  # Adjust the window size as needed
moving_average = df["PressReleaseVolume AR"].rolling(window_size).mean()

volume_stdev = 1.5 * statistics.stdev(df["PressReleaseVolume AR"].dropna())
volume_mean = df["PressReleaseVolume AR"].mean()
volume = df.loc[(df["PressReleaseVolume AR"] >= volume_mean + volume_stdev) | (df["PressReleaseVolume AR"] <= volume_mean - volume_stdev)]
x0 = volume_mean - volume_stdev
x1 = volume_mean + volume_stdev
volume_color = "#C44E52"

plt.figure(figsize=(8, 6))
plt.title("Press-Release\nDistribution of News Volume Abnormal Returns\n(One And a Half Standard Deviations Above or Below the Mean)", fontsize=16)
plt.xlabel("News Volume Abnormal Returns", fontsize=14)
plt.ylabel("Density", fontsize=14)

# Plot moving average
plt.plot(df['PressReleaseVolume AR'], moving_average, color='#FF9800', label='Moving Average')

ax = sns.distplot(df['PressReleaseVolume AR'], hist=True, kde=True, color='#388E3C', kde_kws={'linewidth': 2})
kde_x, kde_y = ax.lines[0].get_data()

p1 = plt.axvline(x=x0, color=volume_color, label="Mean $\pm$ 2 SD", linewidth=2)
p2 = plt.axvline(x=x1, color=volume_color, linewidth=2)
ax.fill_between(kde_x, kde_y, where=(kde_x < x0) | (kde_x > x1), interpolate=True, color=volume_color, alpha=0.5)
plt.legend(fontsize=12)

sns.despine()

plt.show()
```
