```python
import pandas as pd
import numpy as np

df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/dfMinute.pkl")
```

```python
df.sort_values(by="Sentiment return")["Sentiment return"].head(10)
```

```python
df = df.loc[(df["Sentiment return"]>-7) & (df["Sentiment return"]<7)]
```

```python
df["Sentiment return"]
```

```python
df.sort_values(by="Sentiment return")["Sentiment return"].head(10)
```

```python
df = df[["VIX","Sentiment","VIX return", "Sentiment return"]]
```

```python
df["VIX NR"] = df["VIX return"].mean()
df["VIX AR"] = df["VIX return"] - df["VIX NR"]

df["Sentiment NR"] = df["Sentiment return"].mean()
df["Sentiment AR"] = df["Sentiment return"] - df["Sentiment NR"]
```

# STDEV

```python
import statistics
vix_stdev = statistics.stdev(df["VIX AR"].dropna())
vix_mean = df["VIX AR"].mean()
len(df.loc[(df["VIX AR"] > vix_mean+vix_stdev) | (df["VIX AR"] < vix_mean-vix_stdev)])
```

```python
vix = df.loc[(df["VIX AR"] > vix_mean+vix_stdev) | (df["VIX AR"] < vix_mean-vix_stdev)]
```

```python
df.sort_values(by="Sentiment AR")
```

```python
statistics.stdev(df["Sentiment AR"].dropna())
```

```python
import statistics
sent_stdev = statistics.stdev(df["Sentiment AR"].dropna())
sent_mean = df["Sentiment AR"].mean()
len(df.loc[(df["Sentiment AR"] > sent_mean+sent_stdev) | (df["Sentiment AR"] < sent_mean-sent_stdev)])
```

```python
sent = df.loc[(df["Sentiment AR"] >= sent_mean+sent_stdev) | (df["Sentiment AR"] <= sent_mean-sent_stdev)]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(font_scale=1.5)
plt.tight_layout(1)
figure, axes = plt.subplots(1, 2,# sharex=True,
                            figsize=(17, 7))
figure.suptitle('Press Release\nOne Standard Diviations Above or below the mean AR')
# axes[0].set_title('first chart with no data')
# axes[1].set_title('second chart with no data')

ax0 = sns.distplot(df['Sentiment AR'], hist=True,color='#388E3C',ax=axes[0])
kde_x, kde_y = ax0.lines[0].get_data()

x0 = sent_mean-sent_stdev
x1 =sent_mean+sent_stdev
p1 = axes[0].axvline(x=x0,color='#EF9A9A')
p2 = axes[0].axvline(x=x1,color='#EF9A9A')
# plt.title("Press Release")

ax0.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
####################################################################


ax1 = sns.distplot(df['VIX AR'], hist=True,color='#388E3C',ax=axes[1])
kde_x, kde_y = ax1.lines[0].get_data()

x2 = vix_mean-vix_stdev
x3 =vix_mean+vix_stdev
#plotting the two lines
p1 = axes[1].axvline(x=x2,color='#EF9A9A')
p2 = axes[1].axvline(x=x3,color='#EF9A9A')
# plt.title("Press Release\nOne Standard diviations Above or below the mean AR")

ax1.fill_between(kde_x, kde_y, where=(kde_x<x2) | (kde_x>x3) , 
                interpolate=True, color='#EF9A9A')
```
