```python
df["VIX NR"] = df["VIX return"].mean()
df["Sentiment NR"] = df["Sentiment return"].mean()

df["VIX AR"] = df["VIX return"] - df["VIX NR"]
df["Sentiment AR"] = df["Sentiment return"] - df["Sentiment NR"]
```

```python
df[["VIX","VIX AR","Sentiment AR"]]
```

# STDEV

```python
import statistics
vix_stdev = 2*statistics.stdev(df["VIX AR"].dropna())
vix_mean = df["VIX AR"].mean()
len(df.loc[(df["VIX AR"] > vix_mean+vix_stdev) | (df["VIX AR"] < vix_mean-vix_stdev)])
```

```python
vix = df.loc[(df["VIX AR"] > vix_mean+vix_stdev) | (df["VIX AR"] < vix_mean-vix_stdev)]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(font_scale=1.5)

ax = sns.distplot(df['VIX AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()

x0 = vix_mean-vix_stdev
x1 =vix_mean+vix_stdev
#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')
plt.title("Press Release\nTwo Standard diviations\nabove or below the mean AR")

ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
```

```python
import statistics
sent_stdev = 2*statistics.stdev(df["Sentiment AR"].dropna())
sent_mean = df["Sentiment AR"].mean()
len(df.loc[(df["Sentiment AR"] > sent_mean+sent_stdev) | (df["Sentiment AR"] < sent_mean-sent_stdev)])
```

```python
sent = df.loc[(df["Sentiment AR"] >= sent_mean+sent_stdev) | (df["Sentiment AR"] <= sent_mean-sent_stdev)]
```

```python
import matplotlib.pyplot as plt
sns.set(font_scale=1.5)

import seaborn as sns
ax = sns.distplot(df['Sentiment AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()

x0 = sent_mean-sent_stdev
x1 =sent_mean+sent_stdev
#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')
plt.title("Minute Press Release\nTwo Standard diviations\nAbove or below the mean AR")

ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
```

```python
vix_set = set(vix.index.to_list())
sent_set = set(sent.index.to_list())
```
