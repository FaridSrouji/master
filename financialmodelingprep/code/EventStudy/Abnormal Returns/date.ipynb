```python
import pandas as pd
import numpy as np

df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/df.pkl")
```

```python
df = df[["VIX","Sentiment","VIX return", "Sentiment return"]]
```

```python
df.loc[df["Sentiment return"]>2]
```

```python
df["VIX NR"] = df["VIX return"].mean()
df["Sentiment NR"] = df["Sentiment return"].mean()
```

```python
df["VIX AR"] = df["VIX return"] - df["VIX NR"]
df["Sentiment AR"] = df["Sentiment return"] - df["Sentiment NR"]
```

```python
df[["VIX AR","Sentiment AR"]]
```

```python
vix_set = set(df["VIX AR"].nlargest(int(.2*len(df))).index)
```

```python
sent_set = set(df["Sentiment AR"].nlargest(int(.2*len(df))).index)
# sent_set = sent_set.index.to_list()
```

```python
len(sent_set.intersection(vix_set))
```

```python
import random
max = df["Sentiment AR"].max()
min = df["Sentiment AR"].min()
randoms = []
for i in range(482):
    randoms.append(random.uniform(max, min))
```

```python
df["randoms"]=randoms
randoms_set = set(df["randoms"].nlargest(int(.2*len(df))).index)
```
