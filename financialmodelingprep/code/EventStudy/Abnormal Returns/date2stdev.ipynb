```python
import pandas as pd
import numpy as np
import seaborn as sns
import statistics
import matplotlib.pyplot as plt
sns.set(font_scale=1.5)

Type="Quantile"
df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/df.pkl")
```

```python
df = df[["VIX","Sentiment","VIX return", "MarketCap"]]
```

```python
df["VIX return"].mean()
```

```python
df["VIX NR"] = df["VIX"].mean()
df["Sentiment NR"] = df["MarketCap"].mean()

df["VIX AR"] = df["VIX"] - df["VIX NR"]
df["MarketCap AR"] = df["MarketCap"] - df["Sentiment NR"]
```

```python
df["VIX AR"].to_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/code/EventStudy/PKL/VIX_AR.pkl")
```

# 1) STDEV

```python
vix_stdev = 2*statistics.stdev(df["VIX AR"].dropna())
vix_mean = df["VIX AR"].mean()
vix = df.loc[(df["VIX AR"] >= vix_mean+vix_stdev) | (df["VIX AR"] <= vix_mean-vix_stdev)]
x0 = vix_mean-vix_stdev
x1 =vix_mean+vix_stdev
plt.title("Press Release\nOne Standard diviations\nabove or below the mean AR")
# PLOT:
len(vix)

ax = sns.distplot(df['VIX AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()

#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')

ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
```

```python
sent_stdev =  2*statistics.stdev(df["MarketCap AR"].dropna())
sent_mean = df["MarketCap AR"].mean()
sent = df.loc[(df["MarketCap AR"] >= sent_mean+sent_stdev) | (df["MarketCap AR"] <= sent_mean-sent_stdev)]
x0 = sent_mean-sent_stdev
x1 =sent_mean+sent_stdev
plt.title("Press Release\nOne Standard diviations\nabove or below the mean AR")
len(sent)
# PLOT:
len(vix)
sns.set(font_scale=1.5)
ax = sns.distplot(df['MarketCap AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()
#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')
ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
            interpolate=True, color='#EF9A9A')
```

```python
vix_set = set(vix.index.to_list())
sent_set = set(sent.index.to_list())
len(vix)
len(sent)
len(vix_set.intersection(sent_set))
```

# Quantile

```python
# Quantile
top90 = df['VIX AR'].quantile(0.9)
bottom10 = df['VIX AR'].quantile(0.1)
vix = df.loc[(df["VIX AR"] >= top90) | (df["VIX AR"] <= bottom10)]
x0 = bottom10
x1 =top90
plt.title("Press Release\nTop and Bottom 10 percent")
# PLOT:
ax = sns.distplot(df['VIX AR'], hist=True,color='#388E3C')
kde_x, kde_y = ax.lines[0].get_data()
#plotting the two lines
p1 = plt.axvline(x=x0,color='#EF9A9A')
p2 = plt.axvline(x=x1,color='#EF9A9A')

ax.fill_between(kde_x, kde_y, where=(kde_x<x0) | (kde_x>x1) , 
                interpolate=True, color='#EF9A9A')
```
