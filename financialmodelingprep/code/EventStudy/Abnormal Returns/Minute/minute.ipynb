```python
import pandas as pd
import numpy as np

df = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/dfMinute.pkl")
```

```python
df.columns.values
```

```python
df = df[["VIX","Sentiment","VIX return", "Sentiment return"]]
```

```python
df.loc[df["Sentiment return"]>2]
```

```python
df["VIX NR"] = df["VIX return"].mean()
df["Sentiment NR"] = df["Sentiment return"].mean()
```

```python
df["VIX AR"] = df["VIX return"] - df["VIX NR"]
df["Sentiment AR"] = df["Sentiment return"] - df["Sentiment NR"]
```

```python
df[["VIX AR","Sentiment AR"]]
```
