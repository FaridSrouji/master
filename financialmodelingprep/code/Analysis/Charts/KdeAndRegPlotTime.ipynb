```python
df_08 = df_hour_part3.loc[df_hour_part3["Hour"]==8]
df_08["sentiment"] = df_08["sentiment"].apply(np.float)
sentiment_yesterday = df_08["sentiment"].shift(periods=1)
sentiment_yesterday = sentiment_yesterday.interpolate()
df_08["sentiment_ln_return"]=np.log(df_08["sentiment"]/sentiment_yesterday)
sentiment_yesterday = df_08["sentiment"].shift(periods=1)
df_08["sentiment_return"]=((df_08["sentiment"]-sentiment_yesterday)/sentiment_yesterday)
df_08 = df_08.ffill()
df_08["sentiment_return"] = df_08["sentiment_return"].interpolate()
```
