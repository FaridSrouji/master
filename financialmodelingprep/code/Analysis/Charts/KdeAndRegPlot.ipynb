# Reg plot and KDE plot (A kernel density estimate (KDE))

```python
# Columns to plot 
plot_columns = ["sentiment_ln_return(t-1)", "sp500_ln_return", "PressReleaseCount"]
fig, axes = plt.subplots(3,2,figsize=(25,30))
for i,col in enumerate(plot_columns):
    print("i="+str(i))
    print("col="+str(col))

    sns.regplot(col,'vix_ln_return',data=df_part3, ax=axes[i,0]) # regression plot
    sns.kdeplot(df_part3[col],df_part3['vix_ln_return'],shade=True, ax=axes[i,1]) # kernel density estimate plot

fig.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images/KdeAndRegPlot.png", bbox_inches='tight')
```
