# Reg plot and KDE plot (A kernel density estimate (KDE))

```python
sns.relplot(x="sentiment_ln_return", y="vix_ln_return", hue="sp500_ln_return",
            col="subject", col_wrap=5,
            height=3, aspect=.75, linewidth=2.5,
            kind="line", data=df_part3.query("Hour>1"));
```

```python
# Columns to plot 
fig, axes = plt.subplots(24,2,figsize=(30,270))
fig.suptitle("Sentiment", fontsize=30)
fig.subplots_adjust(top=1.88)

fig.tight_layout(pad=5.8)
sns.set(font_scale = 1.5)
# fig = plt.gcf()

for i in range(24):
    print("i="+str(i))
    vix_return = df_part3["vix_ln_return"].loc[df_part3["Hour"]==i]
    sentiment_return = df_part3["sentiment_ln_return"].loc[df_part3["Hour"]==i]
    countObsVix = len(vix_return)
    countObsSentiment = len(sentiment_return)
    print("sentiment_return="+str(countObsVix)+" vix_ln_return="+str(countObsSentiment))
    sns.regplot(sentiment_return,vix_return, data = df_part3, ax=axes[i,0]).set_title("Hour:"+str(i)+"\nTotal Observations:"+str(countObsSentiment), fontsize=20) # regression plot
    sns.kdeplot(sentiment_return,vix_return,data = df_part3,shade=True, ax=axes[i,1]).set_title("Hour:"+str(i)+"\nTotal Observations:"+str(countObsSentiment), fontsize=20) # kernel density estimate plot

 
plt.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images_press_release/KdeAndRegPlot/KdeAndRegPlotHourly_sentiment.png", dpi=216,bbox_inches='tight')
plt.show()
```
