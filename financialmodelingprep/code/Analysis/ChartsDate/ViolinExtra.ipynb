```python
# Draw Plot
fig, axes = plt.subplots(1, 1, figsize=(20,7), dpi= 80)
sns.violinplot(x='Year', y='VIX', data=df_VIX_2010_2021.loc[~df_VIX_2010_2021.Year.isin(["2010", "2019"]), :], ax=axes)
axes.set_title('VIX Year-wise Box Plot\n(The Trend)', fontsize=18); 

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(20,7), dpi= 80)
sns.boxplot(x='Year', y='VIX', data=df_VIX_2010_2021, ax=axes[0])
sns.boxplot(x='Month', y='VIX', data=df_VIX_2010_2021.loc[~df_VIX_2010_2021.Year.isin(["2010", "2019"]), :])
axes[0].set_title('Year-wise Box Plot\n(The Trend)', fontsize=18); 
axes[1].set_title('Month-wise Box Plot\n(The Seasonality)', fontsize=18)
plt.show()
```

```python
fig, axes = plt.subplots(1, 1, figsize=(30,15), dpi= 80)
sns.boxplot(x='Day of Week', y='Open', data=df_VIX_2010_2021, ax=axes)
axes.set_title('Day of Week of week-wise Box Plot\n', fontsize=18)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(30,7), dpi= 80)
sns.violinplot(x='Day of Week', y='VIX', data=df_VIX_2010_2021.loc[~df_VIX_2010_2021.Year.isin([2010, 2019]), :], ax=axes[0])
axes[0].set_title('VIX ADJ CLOSE Day of Week of week-wise Box Plot\n', fontsize=18)
sns.violinplot(x='Day of Week', y='Open', data=df_VIX_2010_2021.loc[~df_VIX_2010_2021.Year.isin([2010, 2019]), :], ax=axes[1])
axes[1].set_title('VIX OPEN Day of Week of week-wise Box Plot\n', fontsize=18)
plt.show()
```
