```python
# POSITIVE
plt.rcParams["figure.figsize"] = (12,12)

def line(x,y):
    
    """Fit a line in a scatter 
    based on slope and intercept"""
    
    slope, intercept = np.polyfit(x, y, 1)
    line_values = [slope * i + intercept for i in x]
    plt.plot(x, line_values, 'b')
    
plt.scatter(df_part3["VIX Natural Log"].values,df_part3["Positive Natural Log"].values, 
            color='lightsalmon')

plt.title('Linear Regression Positive Natural Log and VIX Natural Log')
plt.xlabel('VIX Natural Log')
plt.ylabel('Positive Sentiment Natural Log')
a = line(df_part3["VIX Natural Log"].values,df_part3["Positive Natural Log"].values)

plt.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images_press_release/LinearRegression/positive_vix_natural_log.png", bbox_inches='tight')
plt.show()
```
