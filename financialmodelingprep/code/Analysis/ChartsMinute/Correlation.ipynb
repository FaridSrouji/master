# HeatMap

```python
corrmat = df_values.corr()
  
f, ax = plt.subplots(figsize =(20, 18))

sns.heatmap(corrmat, ax = ax, cmap='coolwarm',vmin=-1,vmax=1, annot=True, linewidths = 0.1)
f.tight_layout()
f.suptitle("Press Release\n",fontsize=26,weight='bold')
f.subplots_adjust(top=0.95)
# sns.heatmap(corrmat, ax = ax, cmap ="YlGnBu", annot=True, linewidths = 0.1)
f.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images_press_release/Correlation/Minute/heatmap_values.png", bbox_inches='tight')
f.show()
```

```python
corrmat = df_ln_returns.corr()
f, ax = plt.subplots(figsize =(20, 18))
sns.heatmap(corrmat, ax = ax, cmap='coolwarm',vmin=-1,vmax=1, annot=True, linewidths = 0.1)

# sns.heatmap(corrmat, ax = ax, cmap ="YlGnBu", annot=True,  linewidths = 0.1)
f.tight_layout()
f.suptitle("Press Release\n",fontsize=26,weight='bold')
f.subplots_adjust(top=0.95)
f.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images_press_release/Correlation/Minute/heatmap_ln_returns.png", bbox_inches='tight')
f.show()
```

# Grid Correlation Matrix

```python
corrmat = df_values.corr()
cg = sns.clustermap(corrmat, cmap='coolwarm',vmin=-1,vmax=1, annot=True, linewidths = 0.1, figsize=(20, 20));
# cg = sns.clustermap(corrmat, cmap ="YlGnBu",  annot=True, linewidths = 0.1, figsize=(20, 20));
plt.setp(cg.ax_heatmap.yaxis.get_majorticklabels(), rotation = 0)
plt.tight_layout()
plt.suptitle("Press Release\nCluster Map",fontsize=26,weight='bold')
plt.subplots_adjust(top=0.95)
cg.savefig("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/images_press_release/Correlation/Minute/clustermap_values.png", bbox_inches='tight')
# cg.show()
```
