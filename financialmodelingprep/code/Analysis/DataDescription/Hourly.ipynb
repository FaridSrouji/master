```python
df_vix = pd.read_pickle("/Users/Farid Srouji/Documents/Research/master/financialmodelingprep/data/pkl/hourlyvix_11Oct_31Dec.pkl")
import datetime, pytz, holidays
df_vix["Hour"] = df_vix["date"].apply(lambda x:(datetime.datetime.strptime(x, "%Y-%m-%d %H:%M:%S").hour))
df_vix["Date"] = df_vix["date"].apply(lambda x:pd.to_datetime((datetime.datetime.strptime(x, "%Y-%m-%d %H:%M:%S").date())))
```

```python
df_sentiment.sort_values(by="Date").tail(50)#.values.mean()
```

```python
df = df_vix.merge(df_sentiment,how="left",right_on=["Date","Hour"], left_on=["Date","Hour"])
```

```python
df.index = df["date"]
```

```python
df["Sentiment(t-1)"] = df["Sentiment"].shift(periods=1)
df["Sentiment(t-2)"] = df["Sentiment"].shift(periods=2)
df["close(t-1)"] = df["close"].shift(periods=1)
df["close(t-2)"] = df["close"].shift(periods=2)

df["Sentiment(t-1)"] = df["Sentiment(t-1)"].ffill()
df["Sentiment(t-2)"] = df["Sentiment(t-2)"].ffill()
df["Sentiment_ln_return"]=np.log(df["Sentiment(t-1)"]/df["Sentiment(t-2)"])
df["close_ln_return"]=np.log(df["close(t-1)"]/df["close(t-2)"])*100
```
