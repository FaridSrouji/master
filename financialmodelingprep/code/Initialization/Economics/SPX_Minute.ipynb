```python
#!/usr/bin/env python

try:
    # For Python 3.0 and later
    from urllib.request import urlopen
except ImportError:
    # Fall back to Python 2's urllib2
    from urllib2 import urlopen

import certifi
import json

def get_jsonparsed_data(url):
    """
    Receive the content of ``url``, parse it as JSON and return the object.

    Parameters
    ----------
    url : str

    Returns
    -------
    dict
    """
    response = urlopen(url, cafile=certifi.where())
    data = response.read().decode("utf-8")
    return json.loads(data)

url = ("https://financialmodelingprep.com/api/v3/historical-chart-full/index/1min/%5EGSPC?apikey=8941e8db3f939429e07397564861bc1a")
```

```python
import pandas as pd
rows_list = []

spx = get_jsonparsed_data(url)
file_str_date = '/Users/Farid Srouji/Documents/Research/master/Data/JSON/SPX.json'
f_date = open(file_str_date, "w", encoding='utf-8')
f_date.write(str(spx))
f_date.close()
# print(spx)
```

```python
for one_spx in spx:
    print(one_spx)
    open = "{:.2f}".format(one_spx['open'])
    low = "{:.2f}".format(one_spx['low'])
    high = "{:.2f}".format(one_spx['high'])
    close = "{:.2f}".format(one_spx['close'])
    volume = one_spx['volume']
    new_row = {"date":one_spx['date'], 
               "open":one_spx['open'], 
               "low":one_spx['low'],
               "high":one_spx['high'], 
               "close":one_spx['close'], 
               "volume":one_spx['volume']}
    rows_list.append(new_row)
```
